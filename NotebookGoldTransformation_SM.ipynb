### **GOLD LAYER TRANSFORMATIONS & AGGREGATIONS : SHOPPINGMART DATA**

```python
from pyspark.sql.functions import *

orders_df = spark.read.parquet("Files/ShoppingMart_Silver_Orders/ShoppingMart_OrderData")
reviews_df = spark.read.parquet("Files/ShoppingMart_Silver_Reviews/ShoppingMart_ReviewsData")
socialmedia_df = spark.read.parquet("Files/ShoppingMart_Silver_SocialMedia/ShoppingMart_SocialMediaData")
web_logs_df = spark.read.parquet("Files/ShoppingMart_Silver_web_logs/ShoppingMart_web_logsData")

display(web_logs_df)
```

```python
# Aggregates web log data on user engagement per user on each page and action

# web_logs_df = spark.read.parquet("Files/ShoppingMart_Silver_web_logs/ShoppingMart_web_logsData")
web_logs_df = web_logs_df.groupBy("action","page","timestamp","user_id").count()

web_logs_df.write.mode("overwrite").parquet("Files/ShoppingMart_Gold_web_logs/ShoppingMart_web_logsData")
display(web_logs_df)
```

```python
# Aggregates product reviews to calculate the Average rating of each product

# reviews_df = spark.read.parquet("Files/ShoppingMart_Silver_Reviews/ShoppingMart_ReviewsData")
reviews_df  = reviews_df.groupBy("product_id").agg(avg("rating").alias("avg_rating"))
reviews_df = reviews_df.orderBy(desc("avg_rating"))

reviews_df.write.mode("overwrite").parquet("Files/ShoppingMart_Gold_Reviews/ShoppingMart_ReviewsData")

# display(reviews_df)
```

```python
# Aggregates data for tracking sentiments across different social media platforms

# socialmedia_df = spark.read.parquet("Files/ShoppingMart_Silver_SocialMedia/ShoppingMart_SocialMediaData")

socialmedia_df = socialmedia_df.groupBy("platform","sentiment").count()
socialmedia_df = socialmedia_df.orderBy("platform")
socialmedia_df.write.mode("overwrite").parquet("Files/ShoppingMart_Gold_SocialMedia/ShoppingMart_SocialMediaData")

# display(socialmedia_df)
```
